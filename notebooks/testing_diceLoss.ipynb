```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import segmentation_models_pytorch as smp


def get_loss(name, **loss_kwargs):
    loss = dict(
        cross_entropy=nn.CrossEntropyLoss,
        dice_loss=smp.losses.DiceLoss,
        focal_loss=smp.losses.FocalLoss,
    )
    return loss[name](**loss_kwargs)


logits = torch.randn(16, 6, 256, 256)  # (batch_size, n_classes, height, width)
masks = torch.randint(0, 6, (16, 256, 256))  # (batch_size, height, width)

loss_kwargs = {"mode": "multiclass"}
# criterion = get_loss("focal_loss", **loss_kwargs)
criterion = get_loss("cross_entropy")
loss = criterion(logits, masks)
loss.item()
```

```python
def get_model(name, **model_kwargs):
    model = dict(
        unet=smp.Unet,
    )
    return model[name](**model_kwargs)


get_model(
    "unet",
    encoder_name="resnet34",
    encoder_weights="imagenet",
    in_channels=3,
    classes=3,
)
```
